*Arbitrary arguments *args and Arbitrary Keyword Arguments **kwargs allow functions to accept a unknown number of arguments.

```python
def my_function(*args): #*args allows for variable number of positional arguments
  print("Type:", type(args)) #it will print the type of args which is a tuple
  print("First argument:", args[0]) #accessing first argument
  print("Second argument:", args[1]) #accessing second argument
  print("All arguments:", args) #this will print all arguments passed

my_function("Emil", "Tobias", "Linus") 
```

```python
def my_function(*numbers): 
  total = 0
  for num in numbers:
    total += num #this calculates the sum of any number of values
  return total

print(my_function(1, 2, 3)) #here 1, 2, 3 are passed as arguments
print(my_function(10, 20, 30, 40))
print(my_function(5))
```

*Function can be used to find the maximum value

```python
def my_function(*numbers): #allows the function to accept any number of positional arguments (packed into a tuple)
  if len(numbers) == 0: #checks if no arguments were passed; if so, returns None
    return None
  max_num = numbers[0] #this initializes the first element as the current maximum
  for num in numbers: #iterates through each number and updates max_num if a larger number is found
    if num > max_num:
      max_num = num
  return max_num #this returns the largest number among the arguments

print(my_function(3, 7, 2, 9, 1)) 
```

*We can use **kwargs to accept any number of keyword arguments

```python
def my_function(**kid): #The double asterisks ** in the parameter mean the function can accept any number of keyword arguments.
  print("His last name is " + kid["lname"]) 

my_function(fname = "Tobias", lname = "Refsnes")
```

*We can combine regular parameters with **kwargs. Regular parameters must come before **kwargs.

```python
def my_function(username, **details): #here username is a regular parameter, **details allows for variable number of keyword arguments
  print("Username:", username) #this prints the username
  print("Additional details:") #this prints a header for additional details
  for key, value in details.items(): #this iterates through each key-value pair in the details dictionary
    print(" ", key + ":", value)

my_function("emil123", age = 25, city = "Oslo", hobby = "coding") 
```

*We can use both *args and **kwargs in the same function. The order must be 1) Regular parameters 2) *args 3) **kwargs

```python
def my_function(title, *args, **kwargs):
  print("Title:", title) #grabs the first argument
  print("Positional arguments:", args) #grabs any remaining positional arguments
  print("Keyword arguments:", kwargs) #grabs any keyword arguments (those with = signs)

my_function("User Info", "Emil", "Tobias", age = 25, city = "Oslo") 
```

#PYTHON SCOPE:

*The LEGB rule-

L refers to Local i.e. inside the current function, E refers to Enclosing i.e. inside enclosing functions (from inner to outer), G refers to Global i.e. at the top level of the module & B refers to Built-in i.e. in Python's built-in namespace



```python
x = "global"

def outer(): #this defines a function outer() that creates a new scope
  x = "enclosing" #here the variable x is local to outer(), but not global
  #inside outer function we define another function
  def inner():
    x = "local"
    print("Inner:", x)
  inner()
  print("Outer:", x) #print this after inner() is called

outer()
print("Global:", x) 
```

```python
def changecase(func): #here changecase is a decorator that takes a function as an argument
  def myinner():
    return func().upper() #here myinner calls the original function and converts its result to uppercase
  return myinner

@changecase #here we apply the changecase decorator to myfunction
def myfunction():
  return "Hello Sally"

@changecase #here we apply the changecase decorator to otherfunction
def otherfunction():
  return "I am speed!"

print(myfunction())
print(otherfunction())
```

```python
def changecase(n):
  def changecase(func):
    def myinner():
      if n == 1: #checking if the argument 'n' passed to changecase() equals 1
        a = func().lower() #calls func() if the above is true and converts its result to lowercase
      else:
        a = func().upper() #calls func() and converts its result to uppercase if n is not 1
      return a
    return myinner
  return changecase

@changecase(1) #here we apply the changecase decorator with argument 1 to myfunction
def myfunction():
  return "Hello Linus" #will print in lowercase because of the decorator argument because of the decorator argument n=1

print(myfunction())
```

*We can use multiple decorators on one function.

-This is done by placing the decorator calls on top of each other.

-Decorators are called in the reverse order, starting with the one closest to the function.

```python
def changecase(func):
  def myinner():
    return func().upper() #here myinner calls the original function and converts its result to uppercase
  return myinner

def addgreeting(func):
  def myinner():
    return "Hello " + func() + " Have a good day!"
  return myinner

@changecase
@addgreeting #using two decorators
def myfunction():
  return "Tobias"

print(myfunction())
```

*Python has a built-in function called functools.wraps that can be used to preserve the original function's name and docstring

```python
import functools

def changecase(func):
  @functools.wraps(func) #this preserves the original function's metadata
  def myinner():
    return func().upper()
  return myinner

@changecase
def myfunction():
  return "Have a great day!"

print(myfunction.__name__) #this will print 'myfunction' instead of 'myinner' because of functools.wraps.
#__name is an attribute that stores the name of the function
```

*Recursion is a common mathematical and programming concept. It means that a function calls itself. This has the benefit of meaning that you can loop through data to reach a result. 

```python
def factorial(n):
  # Base case
  if n == 0 or n == 1:
    return 1
  # Recursive case
  else:
    return n * factorial(n - 1)

print(factorial(5)) 
```

*Python generators are functions that can pause and resume the execution. The code inside the function is only compiled and not executed.

```python
def count_up_to(n):
  count = 1
  while count <= n:
    yield count #the yield statement pauses the function and returns the current value of count
    count += 1

for num in count_up_to(5):
  print(num) 
```

```python
def fibonacci():
  a, b = 0, 1
  while True:
    yield a
    a, b = b, a + b

# Get first 100 Fibonacci numbers
gen = fibonacci()
for _ in range(100): #It can continue generating values indefinitely, without running out of memory
  print(next(gen)) 
```

```python
def echo_generator(): #here we define a generator function
  while True:
    received = yield #the yield expression allows the generator to receive a value when resumed
    print("Received:", received)

gen = echo_generator()
next(gen) #prime the generator
gen.send("Hello") #The send() method allows to send a value to the generator
gen.send("World")
```

```python
def my_gen():
  try:
    yield 1
    yield 2
    yield 3
  finally:
    print("Generator closed")

gen = my_gen()
print(next(gen))
gen.close() #The close() method stops the generator
```

#PYTHON ITERATORS

*To create an object or class as an iterator we have to implement the methods '\_iter\_()' and '\_next\_()' to our object

*Like the function '\_init\_()' which allows us to do some initializing when the object is being created, we use '\_iter\_()' to do operations which returns the iterator object itself

*The '\_next\_()' allows us to do operations which return the next item in the sequence

```python
class MyNumbers:
  def __iter__(self): #this fines the iterator object itself
    self.a = 1 #initializes the starting value                        
    return self

  def __next__(self): #this defines how to get the next value from the iterator
    x = self.a
    self.a += 1
    return x

myclass = MyNumbers()
myiter = iter(myclass)

print(next(myiter))
print(next(myiter))
print(next(myiter))
print(next(myiter))
print(next(myiter))
```

```python
class MyNumbers:
  def __iter__(self):
    self.a = 1
    return self

  def __next__(self):
    if self.a <= 20: #Setting a condition to stop the iteration after a certain point
      x = self.a
      self.a += 1
      return x
    else:
      raise StopIteration #Using the StopIteration statement to prevent the iteration from going on forever

myclass = MyNumbers()
myiter = iter(myclass)

for x in myiter:
  print(x)
```

#Python RegEx or Regular Expression can be used to check if a string contains the specified search pattern

```python
import re  #importing the built-in re module for regular expressions
```

```python
txt = "The rain in Spain"
x = re.search("^The.*Spain$", txt) #searching the string to see if it starts with "The" and ends with "Spain"

if x:
  print("YES! We have a match!")
else:
  print("No match")
```

```python
txt = "The rain in Spain"
x = re.findall("ai", txt) #findall() returns a list of all occurrences of the pattern in the string
print(x) 
```

```python
txt = "The rain in Spain"
x = re.split("\\s", txt) #splitting the string at each whitespace character
print(x)
```

```python
txt = "The rain in Spain"
x = re.sub(r"\s", "9", txt)
print(x) #replacing all whitespace characters with the number 9
```

```python
txt = "The rain in Spain"
x = re.search(r"\bS\w+", txt) #\b indicates a word boundary, S\w+ matches a word starting with 'S'
print(x.span()) #span returns a tuple containing the start and end positions of the match
```

```python
txt = "The rain in Spain"
x = re.search(r"\bS\w+", txt)
print(x.string) #string returns the string passed into the function
```

```python
txt = "The rain in Spain"
x = re.search(r"\bS\w+", txt)
print(x.group()) #group returns the part of the string where there was a match
```

#File Handling:

The key function for working with files in Python is the open() function. The open() function takes two parameters; filename, and mode.

"r" - Read (Default value) Opens a file for reading, error if the file does not exist

"a" - Append Opens a file for appending, creates the file if it does not exist

"w" - Write Opens a file for writing, creates the file if it does not exist

"x" - Create do create the specified file, returns an error if the file exists



```python
f = open("demofile.txt") #this opens the file in read mode by default
print(f.read())
```

```python
f = open("demofile.txt") 
print(f.readline())
f.close() #this closes the file
```

```python
with open("demofile.txt") as f:
  print(f.read(5)) #this reads the first 5 characters of the file
```

```python
#this reads two lines of the file
with open("demofile.txt") as f:
  print(f.readline()) 
  print(f.readline())
```

```python
#Looping through the file object line by line

with open("demofile.txt") as f:
  for x in f:
    print(x)
```
