Global Variables:

*Variables created outside of a function

```python
x = "awesome" #Here x is outside the function so it is a global variable

def myfunc():
  print("Python is " + x)

myfunc()
```

```python
x = "awesome"

def myfunc():
  x = "fantastic" #Here x is within the function so it is a local variable
  print("Python is " + x)

myfunc()

print("Python is " + x)
```

*How to create a global variable inside a function using the global keyword

```python
def myfunc():
  global x #Created a global variable inside a function using the global keyword
  x = "fantastic"

myfunc()

print("Python is " + x)
```

```python
x = "awesome"

def myfunc():
  global x #global keyword is used inside a function to change the value of global variable
  x = "fantastic"

myfunc()

print("Python is " + x)
```

```python
x = "awesome"
print (x) #This will be printed first
def myfunc():
  global x
  x = "fantastic" #Here the global variable gets changed
  print (x)

myfunc()

print(x) #After the function call, the global variable x is still fantastic because the function changed it
```
